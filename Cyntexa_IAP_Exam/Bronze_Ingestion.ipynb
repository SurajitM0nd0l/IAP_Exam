```python
dbutils.widgets.text("catalog", "tpch_dev", "Catalog Name")
catalog = dbutils.widgets.get("catalog")
```

```python
dbutils.widgets.text("pipeline_run_id", "ADHOC", "Pipeline run ID")
pipeline_run_id = dbutils.widgets.get("pipeline_run_id")
```

```python
# Customers table read — from samples.tpch.customer

spark.sql(f"""
CREATE TABLE IF NOT EXISTS IDENTIFIER('{catalog}.bronze.customers')
USING DELTA
AS SELECT *,
       input_file_name() AS _source_file,
       current_timestamp() AS _ingested_at,
       '{pipeline_run_id}' AS _pipeline_run_id
FROM samples.tpch.customer
""")
```
